## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
def marks_processor(marks):
    if len(marks) == 0:
        return None, None, 0, 0

    highest = marks[0]
    lowest = marks[0]
    total = 0

    for mark in marks:
        total += mark
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark

    average = total / len(marks)
    return highest, lowest, total, average

marks = [78, 90, 65, 88, 72]
highest, lowest, total, average = marks_processor(marks)
print("Highest:", highest)
print("Lowest:", lowest)
print("Total:", total)
print("Average:", average)
````

Salary calculation has several smaller steps. Use nested functions to organize the logic cleanly.

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
def process_order(price, quantity, discount_percent=10, tax_percent=5):
    def calculate_subtotal():
        return price * quantity

    def calculate_discount(subtotal):
        return subtotal * discount_percent / 100

    def calculate_tax(amount):
        return amount * tax_percent / 100

    def calculate_final():
        subtotal = calculate_subtotal()
        after_discount = subtotal - calculate_discount(subtotal)
        return after_discount + calculate_tax(after_discount)

    return calculate_final()

final_amount = process_order(1500, 3)
print(f"Final Order Amount: Rs. {final_amount:.2f}")
````

## Question 27 — Employee Payroll System

**Problem:**  
Build a small employee payroll system using multiple functions.

**Requirements:**
Your program should contain **separate functions** for:
- Employee information
- Basic salary
- Allowance
- Deduction

**Rules:**
- Use arguments and `return` properly.
- Use conditions where required (e.g., tax slabs).
- Do **not** write everything inside one function.
- Display a clean final salary report.

**Focus:** Function design + modular programming

````python
def employee_information(name, employee_id):
    return {"name": name, "employee_id": employee_id}

def basic_salary(amount):
    return amount

def allowance(basic):
    return basic * 0.10

def deduction(basic):
    return basic * 0.05

def calculate_tax(taxable_salary):
    if taxable_salary <= 30000:
        return 0
    elif taxable_salary <= 60000:
        return taxable_salary * 0.05
    else:
        return taxable_salary * 0.10

def final_salary(basic):
    gross = basic + allowance(basic)
    total_deduction = deduction(basic)
    tax = calculate_tax(gross - total_deduction)
    return gross - total_deduction - tax, gross, total_deduction, tax

employee = employee_information("Ali", "SMIT-101")
basic = basic_salary(50000)
net, gross, deduction_amount, tax = final_salary(basic)

print("========== EMPLOYEE PAYSLIP ==========")
print("Name:", employee["name"])
print("Employee ID:", employee["employee_id"])
print("Basic Salary: Rs.", basic)
print("Allowance: Rs.", allowance(basic))
print("Gross Salary: Rs.", gross)
print("Deduction: Rs.", deduction_amount)
print("Tax: Rs.", tax)
print("Final Salary: Rs.", net)
print("Note: Tax slabs are sample values for this assignment.")
````

## Question 29 — Student Result Management System

**Problem:**  
Create a complete student result system using functions.

**The system should:**
1. Accept student information
2. Accept multiple subject marks
3. Calculate total
4. Calculate percentage
5. Determine grade
6. Determine pass/fail
7. Display a final result card

**Requirements:**
- Use **multiple functions**
- Use arguments and `return`
- Use lists
- Use conditions and loops
- Use at least one advanced concept (`*args`, `**kwargs`, nested functions, or lambda)
- Do **not** write the entire program inside one function

**Focus:** Combining multiple function concepts

````python
def get_student_info(name, roll_no):
    return {"name": name, "roll_no": roll_no}

def calculate_total(*marks):
    return sum(marks)

def calculate_percentage(total, number_of_subjects, marks_per_subject=100):
    maximum = number_of_subjects * marks_per_subject
    return (total / maximum) * 100 if maximum else 0

def get_grade(percentage):
    if percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"

def get_status(marks):
    return "Pass" if all(mark >= 40 for mark in marks) else "Fail"

def display_result(student, marks):
    total = calculate_total(*marks)
    percentage = calculate_percentage(total, len(marks))
    grade = get_grade(percentage)
    status = get_status(marks)
    subject_names = ["English", "Math", "Computer", "Science"]

    print("========== STUDENT RESULT CARD ==========")
    print("Name:", student["name"])
    print("Roll No:", student["roll_no"])
    for index, mark in enumerate(marks):
        subject = subject_names[index] if index < len(subject_names) else f"Subject {index + 1}"
        print(f"{subject}: {mark}")
    print("Total:", total, "/", len(marks) * 100)
    print(f"Percentage: {percentage:.2f}%")
    print("Grade:", grade)
    print("Status:", status)

student = get_student_info("Hassan", "SMIT-001")
student_marks = [85, 78, 92, 74]
display_result(student, student_marks)

# Advanced concept: *args is used to accept any number of subject marks.
# Lambda could also be used for short calculations, for example:
average = (lambda values: sum(values) / len(values))(student_marks)
print("Average Marks:", average)
````

## Question 30 — FINAL CHALLENGE: Mini E-Commerce Order System

**Problem:**  
Build a small console-based e-commerce order system using Python functions.

The system should simulate a customer placing an order and produce a professional-looking order summary.

**Suggested functions (you decide the exact design):**
- Customer information
- Product selection / product details
- Price & quantity calculation
- Subtotal
- Discount
- Tax
- Delivery charges
- Final bill
- Order summary display

**Mandatory Requirements:**
- Use multiple `def` functions
- Use arguments (including multiple and default where appropriate)
- Use `return`
- Use conditions and loops
- Use lists and/or dictionaries
- Use `*args` **or** `**kwargs`
- Use **at least one nested function**
- Use **at least one lambda** function where it makes sense
- Keep different tasks in separate functions
- Do **not** write the entire program in one function

**Challenge:**  
Design the program yourself. Decide:
- What functions you need
- What arguments each function needs
- What each function should return
- Where conditions, loops, `*args`/`**kwargs`, nested functions, and lambda are useful

This question tests your **complete understanding** of Python functions.

**Focus:** Complete modular real-world system

````python
def customer_information(name, city):
    return {"name": name, "city": city}

def show_products(products):
    print("Available Products:")
    for number, product in enumerate(products, start=1):
        print(f'{number}. {product["name"]} - Rs. {product["price"]}')

def calculate_subtotal(*amounts):
    return sum(amounts)

def calculate_discount(subtotal, percentage=10):
    return subtotal * percentage / 100

def calculate_tax(amount, percentage=5):
    return amount * percentage / 100

def calculate_delivery(subtotal):
    return 0 if subtotal >= 5000 else 250

def create_order(customer, products, selected_items):
    # Nested function: calculate the amount for one selected product.
    def item_total(product, quantity):
        return product["price"] * quantity

    order_lines = []
    totals = []

    for product_index, quantity in selected_items:
        if 0 <= product_index < len(products) and quantity > 0:
            product = products[product_index]
            total = item_total(product, quantity)
            totals.append(total)
            order_lines.append({
                "name": product["name"],
                "price": product["price"],
                "quantity": quantity,
                "total": total
            })

    subtotal = calculate_subtotal(*totals)
    discount = calculate_discount(subtotal)
    after_discount = subtotal - discount
    tax = calculate_tax(after_discount)
    delivery = calculate_delivery(subtotal)
    final_amount = after_discount + tax + delivery

    return {
        "customer": customer,
        "items": order_lines,
        "subtotal": subtotal,
        "discount": discount,
        "tax": tax,
        "delivery": delivery,
        "final_amount": final_amount
    }

def display_order(order):
    print("\n========== E-COMMERCE ORDER SUMMARY ==========")
    print("Customer:", order["customer"]["name"])
    print("City:", order["customer"]["city"])
    print("----------------------------------------------")
    for item in order["items"]:
        print(f'{item["name"]} | Rs. {item["price"]} x {item["quantity"]} = Rs. {item["total"]}')
    print("----------------------------------------------")
    print("Subtotal: Rs.", order["subtotal"])
    print("Discount (10%): Rs.", round(order["discount"], 2))
    print("Tax (5%): Rs.", round(order["tax"], 2))
    print("Delivery: Rs.", order["delivery"])
    print("Final Amount: Rs.", round(order["final_amount"], 2))
    print("==============================================")

products = [
    {"name": "Keyboard", "price": 2500},
    {"name": "Mouse", "price": 1200},
    {"name": "Headphones", "price": 3500},
    {"name": "USB Cable", "price": 500}
]

show_products(products)

customer = customer_information("Hassan", "Karachi")

# Select products by their list index and quantity.
# Example: (0, 1) means 1 Keyboard; (1, 2) means 2 Mice.
selected_items = [(0, 1), (1, 2), (3, 1)]

# Lambda example: convert selected item quantities into a simple list.
quantities = list(map(lambda item: item[1], selected_items))
print("Selected Quantities:", quantities)

order = create_order(customer, products, selected_items)
display_order(order)

# *args is used by calculate_subtotal() for a flexible number of item totals.
# The program uses lists, dictionaries, loops, conditions, return values,
# a nested function, a default argument, and a lambda function.
````
